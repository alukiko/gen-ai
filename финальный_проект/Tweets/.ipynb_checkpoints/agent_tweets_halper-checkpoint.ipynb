# Агент по повышению вовлечённости твитов

В этом ноутбуке реализован интеллектуальный агент, способный анализировать твит, оценивать его предполагаемую вовлечённость и предлагать улучшенные варианты на основе моделей машинного обучения и языковых моделей.



# Функция оценки вовлечённости твита с использованием обученной модели

```python
from typing import Literal
from langchain.tools import tool
import numpy as np
import torch
from transformers import BertTokenizer, BertModel
import joblib
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import nltk

# Один раз при первом запуске
nltk.download("punkt")
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))

# Загрузка модели BERT
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
bert_model = BertModel.from_pretrained("bert-base-uncased")
bert_model.eval()

# Загрузка OneHotEncoder для дня недели
weekday_ohe = joblib.load("weekday_ohe.pkl")

# Загрузка обученной модели RandomForest
rf_model = joblib.load("random_forest_model.pkl")

# Очистка текста
def clean_text(text):
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"[^A-Za-z0-9\s#]", "", text)
    text = re.sub(r"\s+", " ", text).strip().lower()
    tokens = word_tokenize(text)
    return " ".join([w for w in tokens if w not in stop_words])

@tool
def predict_engagement_rf(
    text: str,
    weekday: Literal["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    hour: int
) -> dict:
    """
    Прогнозирует уровень вовлечённости твита по шкале от 0 до 10.
    Использует BERT-эмбеддинги текста, день недели и час публикации.
    
    :param text: Текст твита
    :param weekday: День недели (например, "Monday")
    :param hour: Час публикации (0–23)
    :return: Словарь с предсказанным значением вовлеченности
    """
    # Очистка и BERT-вектор
    cleaned = clean_text(text)
    inputs = tokenizer(cleaned, return_tensors="pt", truncation=True, padding=True, max_length=64)
    with torch.no_grad():
        outputs = bert_model(**inputs)
        bert_vector = outputs.last_hidden_state[:, 0, :].squeeze().numpy().reshape(1, -1)

    # OneHot weekday
    weekday_vector = weekday_ohe.transform([[weekday]])

    # Cyclical time
    hour_sin = np.sin(2 * np.pi * hour / 24)
    hour_cos = np.cos(2 * np.pi * hour / 24)
    hour_vector = np.array([[hour_sin, hour_cos]])

    # Объединение признаков
    features = np.hstack([bert_vector, weekday_vector, hour_vector])

    # Предсказание
    score = rf_model.predict(features)
    return {"engagement_score": round(float(np.clip(score, 0, 10)), 2)}
```

```python
predict_engagement_rf.invoke({
    "text": "Our new product is out. Check the website.",
    "weekday": "Wednesday",
    "hour": 13
})
```

```python
predict_engagement_rf.invoke({
    "text": "Event coming soon. Stay tuned.",
    "weekday": "Wednesday",
    "hour": 13
})
```

# Подключение к LLM модели по API умеющей работать с TOOL для создания ИИ агентов

```python
import json
from langchain_openai import ChatOpenAI
from langgraph.graph import StateGraph, END, MessagesState
from langgraph.prebuilt import ToolNode
from langchain_core.messages import HumanMessage, SystemMessage
from langchain.agents import load_tools
from langchain_core.tools import tool
import os
from langgraph.graph import StateGraph, MessagesState


# Устанавливаем API-ключи
os.environ["VSEGPT_API_KEY"] = ""



# Список инструментов в корректном формате
tools = [predict_engagement_rf]

# Создаем LLM-модель для работы с VseGPT API
llm = ChatOpenAI(
    model="openai/gpt-4o-mini",  # Используем нужную модель VseGPT
    temperature=0.7,
    max_tokens=None,
    openai_api_base="https://api.vsegpt.ru/v1",  # Указываем кастомный URL API
    openai_api_key=os.getenv("VSEGPT_API_KEY"),# Берем API-ключ из переменной окружения
    ).bind_tools(tools)  # <-- Теперь инструменты передаются корректно

print("Модель VseGPT успешно инициализирована!")
```

# Создание простого ИИ агента на Langgraph с функцией оценки качества твита 

```python
# функция, которая определяет нужно ли вызывать инструменты 
# или результат уже получен
def should_continue(state: MessagesState):
    messages = state["messages"]
    last_message = messages[-1]
    if last_message.tool_calls:
        return "tools"
    return END

# функция для ноды взаимодейтсвия с LLM
def gather_data(state: MessagesState):

    messages = state["messages"]

    messages.append(SystemMessage(content='''Системный промт: агент по повышению вовлечённости твитов
Ты — агент по цифровому маркетингу, специализирующийся на увеличении вовлечённости твитов. Твоя задача — сделать твит максимально привлекательным и добиться высокой вовлечённости (score ≥ 8.0).

Ты используешь инструмент predict_engagement_rf, который принимает твит (text), день недели (weekday) и час публикации (hour), и возвращает прогнозируемую вовлечённость по шкале от 0 до 10.

🎯 Что ты делаешь:
Оцениваешь исходный твит через predict_engagement_rf.

Если результат ниже 8.0, переписываешь твит:

усиливаешь посыл,

добавляешь призыв к действию,

корректируешь хештеги,

подбираешь оптимальное время.

подбираешь оптимальный день.

Переоцениваешь улучшенный твит через predict_engagement_rf.

Повторяешь до достижения высокого результата или исчерпания разумных вариантов.

🧪 Инструмент:
predict_engagement_rf(text: str, weekday: str, hour: int) → {"engagement_score": float}

📦 Формат вывода:
📊 Оценка исходного твита:
<число> / 10 — краткий вывод о вовлечённости (низкая / средняя / высокая)

❌ Исходный твит:
"<текст>"

✅ Оптимизированный твит:
"<улучшенный текст>"

📊 Оценка улучшенного твита:
<число> / 10 — оценка от инструмента после оптимизации

🕒 Лучшее время для публикации:
<день недели, час> — основано на исторических паттернах вовлечённости

🏷 Улучшенные хештеги:
<список> — релевантные, специфичные, эффективные

🔁 Если новая версия всё ещё показывает низкую вовлечённость — продолжай итерации до максимального результата или до 3 обращений, после чего выдай лучший результат. Будь креативным, убедительным и ориентированным на результат.
Не обращайся к инструменту больше трех раз за один запрос пользователя
'''))

    response = llm.invoke(messages)

    # информация для отладки
    print(json.dumps(response.tool_calls, indent=2,ensure_ascii=False))
    print(json.dumps(response.content, indent=2,ensure_ascii=False))
   
    return {"messages": [response]}

# Функция сохранения истории сообщений
def add_to_history(state: MessagesState):
    """Добавляет последнее сообщение в историю, если оно есть."""
    messages = state.get("messages", [])
    if messages and isinstance(messages, list):
        state.setdefault("history", []).extend(messages)
    return state
    
# Встроенная в langgraph нода вызова инструментов
tool_node = ToolNode(tools)

# Создаём workflow с памятью
workflow = StateGraph(MessagesState)

# Добавляем узлы
workflow.add_node("gather_data_node", gather_data)
workflow.add_node("tools", tool_node)
workflow.add_node("history", add_to_history)

# Задаём переходы между нодами
workflow.set_entry_point("gather_data_node")
workflow.add_conditional_edges("gather_data_node", should_continue, ["tools", END])
workflow.add_edge("tools", "history")  # Сохраняем сообщения после вызова инструмента
workflow.add_edge("history", "gather_data_node")  # Возвращаемся в цикл общения

# Компилируем граф
graph = workflow.compile()

prompt = ''
while True:
    prompt = input()
    input_messages = [HumanMessage(prompt)]
    output = graph.invoke({"messages": input_messages})
```
